```python
class NumberDescriptor:
    def __set_name__(self, owner, name):
        self.storage_name = f"_{name}"

    def __get__(self, instance, owner):
        if instance is None:
            return self

        return getattr(instance, self.storage_name)

    def __set__(self, instance, value):
        if value > 0:
            setattr(instance, self.storage_name, value)
        else:
            raise ValueError(
                f"{self.storage_name}에는 양수만 입력할 수 있습니다."
            )


class Calculator:
    value1 = NumberDescriptor()
    value2 = NumberDescriptor()

    def __init__(self, value1, value2):
        self.value1 = value1
        self.value2 = value2

    def multiply(self):
        return self.value1 * self.value2
```

```python
calc = Calculator(10, 20)
```

```python
result = calc.multiply()
print(result)
```

```python
# Disrciptor + Data class

from dataclasses import dataclass


class PositiveNumber:
    def __set_name__(self, owner, name):
        self.storage_name = f"_{name}"

    def __get__(self, instance, owner):
        if instance is None:
            return self

        return getattr(instance, self.storage_name)

    def __set__(self, instance, value):
        if value > 0:
            setattr(instance, self.storage_name, value)
        else:
            raise ValueError(
                f"{self.storage_name}에는 양수만 입력할 수 있습니다."
            )


@dataclass
class SensorData:
    temperature: float
    pressure: float

    # Descriptor
    temperature = PositiveNumber()
    pressure = PositiveNumber()


# 객체 생성
sensor = SensorData(25.5, 1013.2)

print(sensor.temperature)
print(sensor.pressure)


'''
데이터 전달 과정

SensorData(25.5, 1013.2)
        │
        ▼
dataclass가 생성한 __init__()
        │
        ├── self.temperature = 25.5
        │          ↓
        │    Descriptor.__set__()
        │          ↓
        │    self._temperature = 25.5
        │
        └── self.pressure = 1013.2
                   ↓
             Descriptor.__set__()
                   ↓
             self._pressure = 1013.2

'''
```

```python
# 값 변경
sensor.temperature = 30.5

print(sensor.temperature)
```

```python
# 잘못된 값 입력
sensor.pressure = -10
```

```python
from dataclasses import dataclass


class PositiveNumber:
    def __set_name__(self, owner, name):
        self.storage_name = f"_{name}"

    def __get__(self, instance, owner):
        if instance is None:
            return self

        return getattr(instance, self.storage_name)

    def __set__(self, instance, value):
        if value > 0:
            setattr(instance, self.storage_name, value)
        else:
            raise ValueError(
                f"{self.storage_name}에는 양수만 입력할 수 있습니다."
            )


@dataclass
class Measurement:
    value1: float
    value2: float

    value1 = PositiveNumber()
    value2 = PositiveNumber()

    def multiply(self):
        return self.value1 * self.value2


measurement = Measurement(10, 20)

print("value1 :", measurement.value1)
print("value2 :", measurement.value2)
print("multiply :", measurement.multiply())

measurement.value1 = 5

print("value1 :", measurement.value1)
print("multiply :", measurement.multiply())
```
