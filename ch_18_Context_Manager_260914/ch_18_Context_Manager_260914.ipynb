```python
# enumerate()를 사용하면 로그의 줄 번호(line number)까지 함께 확인할 수 있습니다. 
# 실제 로그 분석에서는 어느 줄에서 오류가 발생했는지 알 수 있어서 더 유용합니다.

with open("log.txt", "r", encoding="utf-8") as f:

    for line_number, line in enumerate(f, start=1):

        if "error" in line.lower():
            print("⚠️ error를 발견했다")
            print(f"   줄 번호: {line_number}")
            print(f"   로그 내용: {line.strip()}")
            print("-"*50)

        else:
            print("정상이다")
            print("-"*50)
```

```python
import time


class LogAnalyzer:

    def __init__(self, filename):
        self.filename = filename

        # 분석 결과를 저장할 변수
        self.total_lines = 0
        self.error_count = 0
        self.warning_count = 0
        self.info_count = 0

        # 분석 소요 시간
        self.elapsed_time = 0

    # Context Manager 진입
    def __enter__(self):
        print("=== 로그 분석을 시작합니다 ===")

        self.start_time = time.perf_counter()

        return self

    # Context Manager 종료
    def __exit__(self, exc_type, exc_value, traceback):
        self.end_time = time.perf_counter()
        self.elapsed_time = self.end_time - self.start_time

        print("\n=== 로그 분석을 종료합니다 ===")

    # 로그 분석
    def analyze(self):

        with open(self.filename, "r", encoding="utf-8") as f:

            for line_number, line in enumerate(f, start=1):

                self.total_lines += 1

                line = line.lower()

                if "error" in line:
                    self.error_count += 1

                elif "warning" in line:
                    self.warning_count += 1

                elif "info" in line:
                    self.info_count += 1

    # 분석 결과 출력
    def print_result(self):

        print("\n" + "=" * 40)
        print("           로그 분석 결과")
        print("=" * 40)

        print(f"전체 라인 수 : {self.total_lines}")
        print(f"에러 수      : {self.error_count}")
        print(f"경고 수      : {self.warning_count}")
        print(f"INFO 수      : {self.info_count}")

        print("-" * 40)
        print(f"분석 소요 시간 : {self.elapsed_time:.6f}초")

        print("=" * 40)


# ============================================================
# 프로그램 실행
# ============================================================

with LogAnalyzer("log.txt") as analyzer:

    analyzer.analyze()       # 분석이 끝나고 시간 측정 기록이 남아 있음 : 순수한 분석 시간만 기록됨 (260914)
    analyzer.print_result()
```
